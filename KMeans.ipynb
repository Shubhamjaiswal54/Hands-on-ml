```python
df['clusters'] = ypredicted
df.tail()
```

```python
sse = []
krange = range(1 , 11)
for i in krange:
    km = KMeans(n_clusters=i)
    km.fit(df[['PetalLengthCm' , "PetalWidthCm"]])
    sse.append(km.inertia_)

sse
```

```python
#plotting the elbow plot
plt.xlabel("Number of Clusters")
plt.ylabel("sse")
plt.plot(krange ,sse)
plt.grid(True)
plt.show()
```
